# Insight: Top 10 countries with the highest number of recorded cases.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
#worldometer_df = pd.read_csv("/mnt/data/worldometer_data.csv")

# Ensure 'TotalCases' column is numeric
worldometer_df["TotalCases"] = pd.to_numeric(worldometer_df["TotalCases"], errors="coerce")

# Select top 10 countries with highest total cases
top_10_cases = worldometer_df.nlargest(10, "TotalCases")

# Plot the bar chart
plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_cases["Country/Region"], y=top_10_cases["TotalCases"], palette="viridis")

# Formatting the chart
plt.xlabel("Country", fontsize=12)
plt.ylabel("Total Recorded Cases", fontsize=12)
plt.title("Top 10 Countries with Highest Total Cases", fontsize=14)
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```

```python
# Modifying the above code to show label on each bar and revising the Y-Axis label
# Ensure 'TotalCases' column is numeric
worldometer_df["TotalCases"] = pd.to_numeric(worldometer_df["TotalCases"], errors="coerce")

# Select top 10 countries with highest total cases
top_10_cases = worldometer_df.nlargest(10, "TotalCases").copy()

# Convert 'TotalCases' to millions for better readability
top_10_cases["TotalCases_Million"] = top_10_cases["TotalCases"] / 1_000_000

# Plot the bar chart
plt.figure(figsize=(12, 6))
ax = sns.barplot(x=top_10_cases["Country/Region"], y=top_10_cases["TotalCases_Million"], palette="viridis")

# Add labels on each bar
for index, value in enumerate(top_10_cases["TotalCases_Million"]):
    plt.text(index, value + 0.1, f"{value:.2f}M", ha='center', fontsize=12, fontweight='bold')

# Formatting the chart
plt.xlabel("Country", fontsize=12)
plt.ylabel("Total Recorded Cases (in Million)", fontsize=12)
plt.title("Top 10 Countries with Highest Total Cases", fontsize=14)
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```
